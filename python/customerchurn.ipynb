```python
import pyodbc
import pandas as pd

conn = pyodbc.connect(
    'DRIVER={ODBC Driver 17 for SQL Server};'
    'SERVER=LENOVO;'
    'DATABASE=CustChurn;'
    'Trusted_Connection=yes;'
)

print("Connected successfully!")
```

```python
query = "SELECT Contract, Churn FROM dbo.customerchurn;"
df = pd.read_sql(query, conn)
print(df.head())
print(df.shape)
```

```python
contingency_table = pd.crosstab(df['Contract'], df['Churn'])
print(contingency_table)
```

```python
from scipy.stats import chi2_contingency
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-square statistic:", chi2)
print("p-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)


# ============================================================
# Chi-Square Test: Contract Type vs. Churn
# ============================================================
# H0 (null hypothesis): Contract type and Churn are independent 
# (no real relationship; any observed difference is due to chance)
#
# Result: Chi-square statistic = 1184.60, p-value = 5.86e-258
# p-value << 0.05 -> reject the null hypothesis
#
# Conclusion: There is a statistically significant relationship 
# between Contract type and Churn. Month-to-month customers churn 
# at 42.71% vs. 2.83% for two-year contracts (15x gap) — this 
# pattern is essentially certain to be real, not due to random chance.
# ============================================================
```

```python

query2 = "SELECT PaymentMethod, Churn FROM dbo.customerchurn;"
df2 = pd.read_sql(query2, conn)

contingency_table2 = pd.crosstab(df2['PaymentMethod'], df2['Churn'])
print(contingency_table2)

chi2, p_value, dof, expected = chi2_contingency(contingency_table2)
print("Chi-square statistic:", chi2)
print("p-value:", p_value)

# ============================================================
# Chi-Square Test: Payment Method vs. Churn
# ============================================================
# H0 (null hypothesis): Payment method and Churn are independent 
# (no real relationship; any observed difference is due to chance)
#
# Result: Chi-square statistic = 648.14, p-value = 3.68e-140
# p-value << 0.05 -> reject the null hypothesis
#
# Conclusion: There is a statistically significant relationship 
# between Payment method and Churn. Electronic check users churn 
# at 45.29%, roughly 3x every other payment method — this pattern 
# is very unlikely to be due to chance.
# ============================================================

```

```python
query3="SELECT InternetService , Churn FROM dbo.customerchurn;"
df3=pd.read_sql(query3,conn)

contingency_table3=pd.crosstab(df3['InternetService'],df3['Churn'])
print(contingency_table3)

chi2, p_value, dof, expected = chi2_contingency(contingency_table3)
print("Chi-square statistic:", chi2)
print("p-value:", p_value)


# ============================================================
# Chi-Square Test: Internet Service vs. Churn
# ============================================================
# H0 (null hypothesis): Internet service type and Churn are 
# independent (no real relationship; any observed difference 
# is due to chance)
#
# Result: Chi-square statistic = 732.31, p-value = 9.57e-160
# p-value << 0.05 -> reject the null hypothesis
#
# Conclusion: There is a statistically significant relationship 
# between Internet service type and Churn. Fiber optic customers 
# churn at 41.89%, notably higher than DSL (18.96%) and customers 
# with no internet service (7.4%) — this pattern is very unlikely 
# to be due to random chance.
# ============================================================
```

```python
query4 = "SELECT tenure, Churn FROM dbo.customerchurn;"
df4 = pd.read_sql(query4, conn)
df4['tenure_group'] = df4['tenure'].apply(lambda x: 'New (<12 months)' if x < 12 else 'Established')
contingency_table4=pd.crosstab(df4['tenure_group'],df4['Churn'])

chi2, p_value, dof, expected = chi2_contingency(contingency_table4)
print("Chi-square statistic:", chi2)
print("p-value:", p_value)

# ============================================================
# Chi-Square Test: Tenure Group vs. Churn
# ============================================================
# H0 (null hypothesis): Tenure group and Churn are independent 
# (no real relationship; any observed difference is due to chance)
#
# Result: Chi-square statistic = 709.14, p-value = 3.07e-156
# p-value << 0.05 -> reject the null hypothesis
#
# Conclusion: There is a statistically significant relationship 
# between tenure group and Churn. New customers (<12 months) churn 
# at nearly 3x the rate of established customers (48.28% vs. 17.49%) 
# — this pattern is very unlikely to be due to random chance.
# ============================================================

```
